# Kiểm tra dữ liệu trước khi train
Cài package theo README, chuẩn bị frames.jsonl. Overlay để xác minh bbox, class và ignored regions trên dữ liệu thật.

```python
from pathlib import Path
import os
repo = Path.cwd()
if not (repo / 'pyproject.toml').exists():
    repo = repo.parent
os.chdir(repo)
from trafficvision.config import load_config
from trafficvision.data.sources.local import LocalDatasetSource
from trafficvision.data.parsers.xml_parser import CLASS_NAMES
cfg = load_config('configs/data/ua_detrac.yaml')
print(cfg['data'])
```

```python
from PIL import ImageDraw
from IPython.display import display
manifest = Path(cfg['data']['manifest'])
if manifest.exists():
    source = LocalDatasetSource(manifest, 'train')
    image, row = source[0]
    draw = ImageDraw.Draw(image)
    for box, label, track_id in zip(row['boxes'], row['labels'], row['track_ids']):
        draw.rectangle(box, outline='lime', width=2)
        draw.text((box[0], box[1]), f'{CLASS_NAMES[label-1]} #{track_id}', fill='lime')
    for box in row.get('ignored_regions', []):
        draw.rectangle(box, outline='red', width=2)
    print(row['video_id'], row['frame_id'], len(source))
    display(image)
else:
    print('Chạy scripts/prepare_ua_detrac.py hoặc sửa data.manifest trước.')
```
